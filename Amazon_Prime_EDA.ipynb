The analysis includes the distribution of movies and TV shows, content released over different years, popular genres, age certifications, production countries, runtime, IMDb ratings, TMDB ratings, popularity, and IMDb votes.

The project uses univariate, bivariate, and multivariate analysis. Univariate analysis is used to understand individual variables, such as the number of movies and TV shows. Bivariate analysis is used to study relationships between two variables, such as content type and IMDb rating. Multivariate analysis helps in understanding relationships between multiple variables. Different visualizations are used to make these patterns and relationships easier to understand.

The insights obtained from this analysis can help in understanding the overall content available on the streaming platform. Overall, this project demonstrates the complete process of Exploratory Data Analysis, starting from understanding and cleaning raw data to creating visualizations and generating meaningful business insights. ## 3.Data Wrangling

```python
# Remove duplicated from both the dataset
drop_Titles=titles.drop_duplicates(inplace=True)
print(drop_Titles)
drop_credits = credits.drop_duplicates(inplace = True)
print(drop_credits)
```

```python
# Duplicated after removing
print("Titles duplicated:",titles.duplicated().sum())
print("Credits duplicated:",credits.duplicated().sum())
```

```python
# Replace missing age certification with Unknown
titles["age_certification"] = titles["age_certification"].fillna("Unknown")
print(titles["age_certification"])
```

```python
# Count the number of actors for each title
actor_count = credits[credits['role'] == 'ACTOR'].groupby('id').size().reset_index(name='actor_count')
actor_count.head()
```

```python
# Count the number of directors for each title
director_count = credits[credits['role'] == 'DIRECTOR'].groupby('id').size().reset_index(name='director_count')
director_count.head()
```

```python
##+ Merge actor and director counts
credit_summary = pd.merge(actor_count,director_count,on = "id",how="outer")
print(credit_summary)
```

```python
# Replace missing counts with 0
credit_summary["actor_count"] = credit_summary["actor_count"].fillna(0)
credit_summary["director_count"] = credit_summary["director_count"].fillna(0)
print(credit_summary["actor_count"])
print(credit_summary["director_count"])
```

```python
# Merge titles dataset with credit summary
amazon_df = pd.merge(titles,credit_summary,on = "id",how = "left")
(amazon_df).head()
```

```python
# Check final dataset
print("Final dataset:",amazon_df.shape)
```
